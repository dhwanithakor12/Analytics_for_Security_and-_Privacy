#Interpretation of Results : Best Performing Model

Random Forest (Baseline) achieved the highest Accuracy (0.7468) and Recall (0.7468), indicating strong overall predictive capability.

Random Forest + SMOTE improved ROC-AUC, suggesting better class separation under imbalance handling.

SVM (Baseline) demonstrated performance comparable to Random Forest, confirming its effectiveness in capturing non-linear decision boundaries.

Decision Tree models, while interpretable, showed lower predictive performance due to their limited ability to generalize complex patterns.
For discussion purposes, the best-performing variant of each algorithm is
used for cross-model comparison, while baseline and SMOTE variants are included
to analyse the impact of imbalance handling and tuning.

Random Forest Performed Best Because

Random Forest is an ensemble learning method, meaning it combines multiple decision trees trained on different subsets of the data. This reduces overfitting, improves stability, and enhances generalization. The ensemble structure allows Random Forest to model complex, non-linear relationships present in medical data more effectively than a single classifier.

MEDICAL RELEVANCE 
 Recall (Sensitivity) Matters More Than Accuracy Because 

In diabetes prediction, false negatives (classifying a diabetic patient as non-diabetic) pose serious clinical risks:

Delayed diagnosis

Lack of timely treatment

Increased likelihood of complications such as cardiovascular disease, kidney failure, and vision loss

Therefore, Recall (Sensitivity) is prioritized over Accuracy, as it measures the model’s ability to correctly identify diabetic patients.

Clinical Safety Perspective

Random Forest (Baseline) achieved the highest Recall, making it the safest model for healthcare deployment.

SVM + SMOTE improved diabetic class detection but slightly reduced overall stability.

Decision Trees, while interpretable, are less reliable for real-world clinical screening due to lower recall.

From a medical standpoint, Random Forest offers the best balance between detection accuracy and patient safety.
